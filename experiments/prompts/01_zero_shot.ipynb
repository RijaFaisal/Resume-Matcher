```python
%pip install groq
import os
from groq import Groq
from dotenv import load_dotenv
load_dotenv()
api_key = os.getenv("GROQ_API_KEY")
client = Groq(api_key=api_key)
```

```python
zero_shot_prompt = """
You are an expert resume screener.
Given this job description: {JOB}
and this resume: {RESUME}
answer: Should the candidate be shortlisted? Briefly justify.
"""
```

```python
def run_zero_shot(job, resume):
    prompt = zero_shot_prompt.format(JOB=job, RESUME=resume)

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": prompt}],
    )

    return response.choices[0].message.content

print(run_zero_shot("Software Engineer role requiring Python and APIs",
                    "BS CS student with Python, ML, API experience"))
```
