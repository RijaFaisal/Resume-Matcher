```python
%pip install groq
import os
from groq import Groq
from dotenv import load_dotenv
load_dotenv()
api_key = os.getenv("GROQ_API_KEY")
client = Groq(api_key=api_key)
```

```python
cot_system = "You are a careful assistant. Answer step-by-step and then give final verdict."
def run_cot(query):
    messages = [
      {"role":"system","content":cot_system},
      {"role":"user","content": "Please think step-by-step and then answer:" + query}
    ]
    r = client.chat.completions.create(model="llama-3.1-8b-instant", messages=messages)
    return r.choices[0].message.content

print(run_cot("Q: Candidate: '5 years Java' Job: 'Backend engineer (Python preferred)' -> Match?"))
```
